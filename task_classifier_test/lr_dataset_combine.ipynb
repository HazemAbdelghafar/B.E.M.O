```python
# Combine the two DataFrames
df_combined = pd.concat([df, df2], ignore_index=True)
```

```python
# Shuffle the combined DataFrame
df_combined = df_combined.sample(frac=1).reset_index(drop=True)
```

```python
df_combined.info()
```

```python
# Get the sum of numerical columns
general_sum = df_combined['general'].sum()
smart_home_sum = df_combined['smart_home'].sum()
todo_sum = df_combined['todo'].sum()
mail_sum = df_combined['mail'].sum()
learning_resources_sum = df_combined['learning_resources'].sum()

print(f"General: {general_sum}")
print(f"Smart Home: {smart_home_sum}")
print(f"Todo: {todo_sum}")
print(f"Mail: {mail_sum}")
print(f"Learning Resources: {learning_resources_sum}")
```
