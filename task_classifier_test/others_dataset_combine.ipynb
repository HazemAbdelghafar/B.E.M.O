```python
# Get the sum of numerical columns
general_sum = df['general'].sum()
smart_home_sum = df['smart_home'].sum()
todo_sum = df['todo'].sum()
mail_sum = df['mail'].sum()
others_sum = df['others'].sum()

print(f"General: {general_sum}")
print(f"Smart Home: {smart_home_sum}")
print(f"Todo: {todo_sum}")
print(f"Mail: {mail_sum}")
print(f"Others: {others_sum}")
```

```python
# Add learning_resources column
df['learning_resources'] = 0
```

```python
df.sample(5)
```

```python
df2 = pd.read_csv('./datasets/bemo_prompts_dataset_v1.1.csv')
```

```python
df2.sample(5)
```

```python
# Add others column to df2
df2['others'] = 0
```

```python
# Combine the two dataframes
df_combined = pd.concat([df, df2], ignore_index=True)
```

```python
# Get the sum of numerical columns
general_sum = df_combined['general'].sum()
smart_home_sum = df_combined['smart_home'].sum()
todo_sum = df_combined['todo'].sum()
mail_sum = df_combined['mail'].sum()
others_sum = df_combined['others'].sum()
learning_resources_sum = df_combined['learning_resources'].sum()

print(f"General: {general_sum}")
print(f"Smart Home: {smart_home_sum}")
print(f"Todo: {todo_sum}")
print(f"Mail: {mail_sum}")
print(f"Learning Resources: {learning_resources_sum}")
print(f"Others: {others_sum}")
```

```python
# Reorder the columns making the others column the last one
df_combined = df_combined[['prompt', 'general', 'smart_home', 'todo', 'mail', 'learning_resources', 'others']]
```
